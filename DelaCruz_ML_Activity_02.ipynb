```python
df["pass"] = (df["G3"] >= 10).astype(int)
```

```python

print(df["pass"].value_counts())
```

```python
features = [
    "age",
    "studytime",
    "failures",
    "absences",
    "G1",
    "G2"
]

X = df[features]
y = df["pass"]
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size = 0.20,
    random_state = 42,
    stratify=y
)
```

```python
model=DecisionTreeClassifier(
    max_depth=4,
    random_state=42
)

model.fit(X_train, y_train)
```

```python
y_pred = model.predict(X_test)
print(y_pred)
```

```python
accuracy = accuracy_score(y_test, y_pred)
print("Accuracy:", accuracy)
```

```python
cm = confusion_matrix(y_test, y_pred)

disp = ConfusionMatrixDisplay(
    confusion_matrix=cm,
    display_labels=["Fail", "Pass"]
)

disp.plot()

plt.title("Student Pass/Fail Confusion matrix")
plt.show()
```

```python
results = X_test.copy()

results["Actual"] = y_test
results["Predicted"] = y_pred

results.head(10)
```

```python
results["Risk"] = results["Predicted"].map({
    0: "At Risk",
    1: "Likely to Pass"
})

results.head(10)
```

```python
results["Recommendation"]=results["Predicted"].map({
    0: "Prioritize for academic intervention",
    1: "Continue regular monitoring"
})

results.head(10)
```

```python
probabilities = model.predict_proba(X_test)

results["Probability_of_Failing"] = probabilities[:,0]

results.sort_values(
    "Probability_of_Failing",
    ascending=False
).head(20)
```
